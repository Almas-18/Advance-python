```python
#TYPE OF IMPLEMENTATION OF FUNCTION
'''
1. NRNP: No Return, No Parameter
2. NRWP: No Return, With Parameter
3. RWP: Return, With Parameter
4. RWNP: Return, No Parameter
'''
```

```python
#function without argument and withouut return typr
#NRNP
def greet():
    print("Hello,Welcome to function")
greet()
greet()
```

```python
#function without argument and withouut return typr
#NRNP
def add():
    a=int(input("Enter value of a:"))
    b=int(input("Enter value of b:"))
    sum=a+b
    print("Sum is:",sum)
add()
add()
add()
```

```python
#function without argument and withouut return typr
#NRNP
def print_num():
    for i in range(1,11):
        print(i)
print_num()
```

```python
#NO ARGUMENT BUT WITH RETURN 
#NPWR
def welcome():
    return 'welcome to the event'
welcome()
welcome() #the function returns the message 4 times, but it is not automatically displayed in a normal Python script.
```

```python
#NO ARGUMENT BUT WITH RETURN 
#NPWR
def add():
    return 10+20+30
add()
```

```python
#NO ARGUMENT BUT WITH RETURN 
#NPWR
def multiply():
    num=int(input("Enter num:"))
    return num*2
print("Output is::",multiply())
```

```python
#NPWR
def element():
    return[10,20,30,40]
element()
    
```

```python
def Even_Odd():
    num=int(input("Even numberr::"))
    if num%2==0:
        return"Even Number"
    else:
        return"Odd Number"
Even_Odd()
```

```python
#NPWR
#are of rectamgle
def Area_Reactangle():
    len=int(input("Enter len:"))
    width=int(input("Enter width"))
    area=len*width
    return area
Area_Reactangle()
```

```python
#perimeter of rect
def perimeter_rect():
    len=int(input("Enter len:"))
    width=int(input("Enter width"))
    perimetr=2*(len+width)
    return perimetr
perimeter_rect()
```

```python
# simple intrest with NPWR
# - Formula: (SI = p*R*T/100)
def simple_intrest():
    p=1000
    r=5
    t=2
    si=(p*r*t)/100
    return si
simple_intrest()
    
```

```python
#NRwP(with para/argument,no retuttrn)
def add(a,b):
    print("sum is:",a+b)
add(5,3)
add(2,4)
```

```python
#area of circle using NRWP
def Area_circle(r):
    area=3.14*r*r
    print("Area of circle :",area)
# r=float(input("Enter radius::"))
Area_circle(r=float(input("Enter radius: ")))
```

```python
#Even number
def even(n):
    for i in range(2,n+1,2):
        print(i)
even(20)
        
```

```python
# With Argument, With Return.
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact
n=int(input("Enter num:"))
print("Factorial of num is:",factorial(n))
     
```

```python
def max_num(a,b):
    if a>b:
        return a
    return b
print(max_num(10,20))
```

```python
def count_even(numbers):
    even=0
    for i in numbers:
        if i%2==0:
            even=even+1
    return even
      
count_even(numbers = [3, 8, 10, 5, 7, 12])

```
