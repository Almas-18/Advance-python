```python
#FILTER
'''

Filter

* filter() is a built in Python function used to filter the elements from an iterable
  based on some condition
* It returns a filter object, which is usually converted to list or tuple or set
* The condition is specified using a function or a lambda function
* It does not modify the original dataset/iterable
* Syntax : filter(function,iterable)
* function : return true/false
* iterable : list/tuple/set
* In return (output) a filter objects containing only the elements that satisfy the
  condition

'''
```

```python
#even number
num=[1,2,3,4,5,6,7,8]
even=list(filter(lambda n:n%2==0,num))
print(even)
print(num)
```

```python
#odd number
num=[1,2,3,4,4,5,6,7,8]
odd=list(filter(lambda n:n%2!=0,num))
print(odd)
```

```python
#print word which is len>3
words=['hi','hello','saniya','sammina']
Result=list(filter(lambda n:len(n)>3,words))
print(Result)
```

```python
#positive number
num=[1,2,33,-4,-5,-6]
positive=list(filter(lambda n:n>0,num))
print(positive)
```

```python
#negative number
num=[1,2,33,-4,-5,-6]
negative=list(filter(lambda n:n<0,num))
print(negative)
```

```python
#filter that start with A
name=["Amit","Anil",'Asim','saniya','masira']
Result=list(filter(lambda n:n.startswith("A"),name))
print(Result)
```

```python
#filter the vowels from list
letter=['a','p','o','u','e','i']
res=list(filter(lambda l:l in'aioue',letter))
print(res)
```

```python
'''
2. Map():

* map() applies a function to every element of an iterable
* it returns a map object, which is usually converted into list
* it creates a transformed version of the original iterable
* the original iterable remains unchanged
* syntax : map(function,iterable)
* function : function to apply
* iterable : can be list,tuple,etc
* it returns a map object containing transformed values
'''
```

```python
#make square of each element in list
num=[1,2,3,4,5]
res=list(map(lambda n:n**2,num))
print(res)
print(num)
```

```python
#make cube of each element in list
num=[1,2,3,4,5]
res=list(map(lambda n:n**3,num))
print(res)
```

```python
#convert all string to uppercase
name=['Amit',"raj",'soni']
res=list(map(lambda n:n.upper(),name))
print(res)
```

```python
#fine length of all string
name=['Amit',"raj",'soniya']
res=list(map(len,name))
print(res)
```

```python
#convert int to string
num=[10,20,30,40,50]
res=list(map(str,num))
print(res)
```

```python
# 3remove extra space
name=["   Amit","Raj    ","ram  "]
res=list(map(str.strip,name))
print(res)
```

```python
#double eaach num
num=[1,2,3,4,5,6]
res=list(map(lambda n:n*2,num))
print(res)
```

```python
# convert celsius to fahrenheit -- (t*9/5)+32
celsius=[0,10,220,30]
res=list(map(lambda t:(t*9/5)+32,celsius))
print(res)
```

```python
# 3. Multiply each number by 5
num=[5,6,4,3]

res=list(map(lambda n:n*5,num))
print(res)
```

```python
# 4. Add 10 to each number
num=[1,2,3,4,5]
res=list(map(lambda n:n+10,num))
print(res)
```
