```python
#VARIABLE LENGTH ARGUMENT:

def sum(*nums):
    s=0
    for i in nums:
        s+=i
    print(s)
sum(1,2,3)
```

```python
#MAXIMUM NUMBER
def max(*nums):
    max1=nums[0]
    for i in nums:
        if i>max1:
            max1=i
    print(max1)
max(1,9,0,8,3,20,100,800)
```

```python
#6.LAMBDA FUNCTION---> ANONYMOUS FUNCTION
''' 
SYNTAX-->lambda var:True statement if condition else false condition
1>NO FUNCTION NAME
2.USED FOR SIMPLE OPERATION
3.AND ONE LINE FUNCTION
4.VARIABLE WILL BE CONVERTED TO FUNCTION AT RUNTIME
'''
```

```python
#ADD TWO VARIABLE
#VAR=LAMBDA KEYWORD ARGUMENT:OPERATION
add=lambda a,b:a+b
print(add(5,4))
```

```python
#based on second elemen---sorting
l1=[(6,3),(2,1),(4,2)]
l1.sort(key=lambda n:n[1])
print(l1)
```

```python
#filter positive element
l1=[-2,-9,3,9,4,-5,7]

result=list(filter(lambda num:num>0,l1))
print(result)
# filter() checks each element in the list
# lambda num takes one element at a time
# num > 0 checks whether the element is positive
# If True, the element is kept; if False, it is rejected
# list() converts the filter result into a list
```

```python
#convering local var to global var
# Creating a global variable
count = 0

def update():
    # global allows us to modify the global variable inside the function
    global count

    # Increase the global count by 10
    count += 10

# Calling the function
update()

# Printing the updated global variable
print(count)  # Output: 10
```

```python
# 1. To convert local var to global var
# 2. if you want to update global var in local scope use global keyword

balance=5000
print(balance)

def deposit():
    global balance
    balance+=2000  # Add 2000 to the existing balance

deposit()
print("After Deposit the balance is:",(balance))
```

```python
#GLOBAL AND LOCAL VAR (INTERVIEW Q)
num=10
def test():
    # global num
    num=20
    print('Local variable',num)
test()
print('Global variable',num)
```

```python
#RECURSION :FUN CALL ITSELF
def factorial(n):
    if n==1:
        return 1
    return n*factorial(n-1)
print(factorial(5))
```

```python
def sum(n):
    if n==0:
        return 0
    return n+sum(n-1)
print(sum(5))
```
