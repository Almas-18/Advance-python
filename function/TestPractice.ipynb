```python
# Part B: Find the maximum and minimum in a list
def find_values(num):
    maximum = num[0]
    minimum = num[0]

    for i in num:
        if i > maximum:
            maximum = i

        if i < minimum:
            minimum = i

    return maximum, minimum

num = [10, 5, 30, 2, 20]

result = find_values(num)

print("Max =", result[0],"min=",result[1])
# print("Minimum =", minimum)
```

```python
# Part C: Find the product of all numbers in a lis
from functools import reduce

def fact(num):
    product = reduce(lambda a, b: a * b, num)
    print(product)

fact(num=[1, 2, 3, 4, 5])
```

```python
# Q8. Find the factorial of a number using recursion
def fact(n):
    if n==0 or n==1:
        return 1
    return n*fact(n-1)
fact(5)
```

```python
# Q6. Shopping Cart Total Using *args

def calculate_cart_total(*prices):
    total = 0

    for price in prices:
        total += price

    return total

print(calculate_cart_total(100, 200, 300, 400))
```

```python
# Q7. Student Marks Analyzer Using *args

def analyze_marks(*marks):
    total = sum(marks)
    average = total / len(marks)
    highest = max(marks)
    lowest = min(marks)

    print("Total:", total)
    print("Average:", average)
    print("Highest:", highest)
    print("Lowest:", lowest)

analyze_marks(80, 90, 70, 60, 100)
```

```python
# 9. Bank Operations

def bank_operation(balance, operation, amount):
    if operation == "deposit":
        balance = balance + amount

    elif operation == "withdraw":
        if amount <= balance:
            balance = balance - amount
        else:
            return "Insufficient balance"

    else:
        return "Invalid operation"

    return balance


print(bank_operation(5000, "deposit", 2000))
print(bank_operation(5000, "withdraw", 1000))
print(bank_operation(500, "withdraw", 1000))
```

```python
# Q10. Product Discount Calculator Using lambda

discount = lambda price, percent: price * percent / 100

price = 1000
percent = 20

discount_amount = discount(price, percent)
final_price = price - discount_amount

print("Discount Amount:", discount_amount)
print("Final Price:", final_price)
```

```python
# Q11. Password Validation
def validate_password(password):
    if len(password) < 8:
        return False

    upper = False
    lower = False
    digit = False

    for ch in password:
        if ch.isupper():
            upper = True
        elif ch.islower():
            lower = True
        elif ch.isdigit():
            digit = True

    return upper and lower and digit


print(validate_password("Hello123"))
print(validate_password("hello123"))
```

```python

def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n - factorial(n - 1)


print(factorial(5))
```

```python
# ATM transactions
def transaction_summary(opening_balance, *transactions):
    balance = opening_balance

    for amount in transactions:
        balance += amount

    return balance

print(transaction_summary(5000, 1000, -500, 2000))
```

```python
# Q15. Company Salary Report
company_name = "ABC Company"

def calculate_net_salary(basic, allowances=5000, deductions=2000):
    return basic + allowances - deductions


def display_salary_report(**employee):
    print("Company:", company_name)
    for key, value in employee.items():
        print(key, ":", value)


def total_salary_expense(*salaries):
    return sum(salaries)


net_salary = calculate_net_salary(30000)

display_salary_report(
    name="Amit",
    basic=30000,
    net_salary=net_salary
)

print("Total Salary Expense:",
      total_salary_expense(33000, 40000, 35000))
```
