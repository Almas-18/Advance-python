```python
# 4. Area of Circle: Create a function to take the radius as input and calculate the area using \(\pi = 3.14\).
def area():
    radius=int(input("enter radius:"))
    area=3.14*radius*radius
    print("Area of circle:",area)
area()    
```

```python
# 5. Factorial: Create a function to take a number as input and calculate its factorial using a for loop.
def fact():
    num=int(input("enter a number:"))
    fact=1
    for i in range(1,num+1):
        fact=fact*i
    print("Factorial of number is:",fact)
fact()
```

```python
# 6. Area of Rectangle: Create a function to take length and width as input and calculate the area.
def Area_Reactangle():
    length=int(input("Enter a length:"))
    width=int(input("Enter a width:"))
    area=length*width
    print("Area of rect:",area)
Area_Reactangle()
```

```python
# 7. Perimeter of Rectangle: Create a function to take length and width as input and calculate the perimeter.
def perimeter():
    length=int(input("Enter a length:"))
    width=int(input("Enter a width:"))
    perimeter=2*(length+width)
    print("perimeter of rect:",perimeter)
perimeter()
```
