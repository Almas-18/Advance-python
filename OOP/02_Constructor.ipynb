# Dunder Methods in Python

**Definition:** Dunder methods are special methods that start and end with double underscores (`__`). Python calls them automatically for specific operations.

# Common Dunder Methods in Python

| Method | Use |
|---|---|
| `__init__()` | Constructor; initializes an object. |
| `__str__()` | Returns readable text when printing an object. |
| `__repr__()` | Returns a developer-oriented representation of an object. |
| `__len__()` | Defines the result of `len()` for an object. |
| `__add__()` | Defines the `+` operator. |
| `__sub__()` | Defines the `-` operator. |
| `__eq__()` | Defines the `==` comparison. |


### Example: `__str__()`

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

s1 = Student("Ram")
print(s1)
```

**Output:**
```text
Ram
```

**Note:** Dunder means *Double Underscore*. These methods are also called **magic methods** or **special methods**.

````python
#parametrized constructor
class Student:
    def __init__(self,rollno,name,course):
        self.rollno=rollno
        self.name=name
        self.course=course
s1=Student(1,'riya','python')
s2=Student(2,'raj','cs')
print(s1.rollno,s1.name,s1.course)
print(s2.rollno,s2.name,s2.course)

        
````

````python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def display(self):
        print("======= Student Details =======")
        print("Name is:", self.name)
        print("Course is:", self.course)


# User input for first student
n1 = input("Enter Name: ")
c1 = input("Enter course: ")

s1 = Student(n1, c1)
s1.display()

# User input for second student
n2 = input("Enter Name: ")
c2 = input("Enter course: ")

s2 = Student(n2, c2)
s2.display()
````

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total(self):
        return self.price * self.quantity


# User input
name = input("Enter name of product: ")
price = float(input("Enter the price of product: "))
quantity = int(input("Enter the quantity of product: "))

# Create object
p = Product(name, price, quantity)

# Display product details
print("\n======= Product Details =======")
print("Product name is:", p.name)
print("Price of product:", p.price)
print("Quantity of product:", p.quantity)
print("Total cost is:", p.total())
````

````python
# Question: Write a Python program using a Car class and constructor to accept the car name, rent per day, and number of days,
# then calculate and display the total rent.
class Car:
    def __init__(self, name, rentperday, day):
        self.name = name
        self.rentperday = rentperday
        self.day = day

    def total(self):
        return self.rentperday * self.day


name = input("Enter name of car: ")
rentperday = float(input("Enter rent per day: "))
day = int(input("Enter number of days: "))

c1 = Car(name, rentperday, day)

print("\n======= Car Details =======")
print("Car name:", c1.name)
print("Rent per day:", c1.rentperday)
print("Number of days:", c1.day)
print("Total rent:", c1.total())
        
````
