## Encapsulation in Python

**Definition:** Encapsulation means wrapping data (variables) and methods (functions) into a single class and controlling access to the data.

**Simple formula:** Encapsulation = Data Hiding + Controlled Access

### Why Is Encapsulation Important?

1. **Data Protection:** Prevents accidental changes to data.
2. **Controlled Access:** Controls how data is accessed and modified.
3. **Data Validation:** Prevents invalid values from being stored.
4. **Code Reusability & Maintenance:** Makes code easier to manage and update.

### How It Works
1. Store data inside a class.
2. Use access conventions to control how data is accessed.
3. Use methods to validate and update data.

### How Access Modifiers Work in Python

1. **Public (`name`):** Can be accessed from anywhere.
2. **Protected (`_name`):** Intended for use inside the class and its child classes.
3. **Private (`__name`):** Uses name mangling to make direct access harder.

**Note:** Python does not strictly restrict access to protected and private variables.

### Syntax

```python
class Student:
    def __init__(self):
        self.name = "Almas"       # Public
        self._age = 21            # Protected
        self.__marks = 90         # Private
```

### Types of Access Modifiers

| Type | Syntax | Meaning |
|---|---|---|
| Public | `name` | Can be accessed from anywhere. |
| Protected | `_name` | Intended for use inside the class and its subclasses. |
| Private | `__name` | Uses name mangling to make direct access harder. |

**Note:** Python does not strictly enforce protected or private access.

### Simple Real-Life Example

A bank account protects its balance by allowing updates through deposit and withdrawal methods instead of changing the balance directly.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print("Balance:", self.__balance)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount


account = BankAccount(5000)
account.deposit(1000)
account.show_balance()
```

**Output:**
```text
Balance: 6000
```
### Key Points
- Private variables are accessed using getter and setter methods.
- Python does not strictly enforce access modifiers; it follows naming conventions.
- Private variables use name mangling (`_ClassName__variable`).

````python
class Student:
    def __init__(self, name, marks):
        self.name = name       # Public
        self.__marks = marks   # Private

    def get_marks(self):       # Getter: retrieve data
        return self.__marks

    def set_marks(self, m):    # Setter: update data
        if m >= 0:
            self.__marks = m
        else:
            print("Invalid marks")


s = Student('Riya', 90)

print(s.name)
print(s.get_marks())

s.set_marks(95)
print(s.get_marks())
````

````python
# 2. Library Book — Encapsulation
class LibraryBook:
    def __init__(self,book_id,book_name):
        self.__book_id=book_id
        self.__book_name=book_name
        self.__is_issued=False #False? Because when we first add a book to the library, it is available.
    def issue_book(self):
        if not self.__is_issued:
            self.__is_issued=True
            print('Book issued successfully!')
        else:
            print('Book is already issued!')
    def return_book(self):
        if self.__is_issued:
            self.__is_issued=False
            print('Book returned successfully!')
        else:
            print('Book is not issued!')
    def display_status(self):
        if self.__is_issued:
            self.__is_issued=True
            print('Book is issued')
        else:
            print('Book is available')
b=LibraryBook(101,'java')
b.display_status()

b.issue_book()
b.display_status()
        
b.return_book()
b.display_status()
````

````python

# ATM PIN SECURITY SYSTEM

class ATM:
    def __init__(self, account_number, balance, pin):
        self.__account_number = account_number
        self.__balance = balance
        self.__pin = str(pin)

    def verify_pin(self):
        entered_pin = input("Enter the PIN: ")

        if entered_pin == self.__pin:
            print("PIN verified successfully!")
        else:
            print("Incorrect PIN!")

    def check_balance(self):
        entered_pin = input("Enter the PIN: ")

        if entered_pin == self.__pin:
            print("Your balance is:", self.__balance)
        else:
            print("Incorrect PIN!")

    def withdraw(self):
        entered_pin = input("Enter the PIN: ")

        if entered_pin == self.__pin:
            amount = float(input("Enter withdrawal amount: "))

            if amount > 0:
                if amount <= self.__balance:
                    self.__balance -= amount
                    print("Withdrawal successful!")
                    print("Remaining balance:", self.__balance)
                else:
                    print("Insufficient balance!")
            else:
                print("Amount must be greater than zero!")
        else:
            print("Incorrect PIN!")

    def deposit(self):
        entered_pin = input("Enter the PIN: ")

        if entered_pin == self.__pin:
            amount = float(input("Enter deposit amount: "))

            if amount > 0:
                self.__balance += amount
                print("Deposit successful!")
                print("Updated balance:", self.__balance)
            else:
                print("Amount must be greater than zero!")
        else:
            print("Incorrect PIN!")

    def change_pin(self):
        old_pin = input("Enter old PIN: ")

        if old_pin == self.__pin:
            new_pin = input("Enter new PIN: ")

            if len(new_pin) == 4 and new_pin.isdigit():
                self.__pin = new_pin
                print("PIN changed successfully!")
            else:
                print("PIN must contain exactly 4 digits!")
        else:
            print("Incorrect old PIN!")

    def display_details(self):
        entered_pin = input("Enter the PIN: ")

        if entered_pin == self.__pin:
            print("Account number:", self.__account_number)
            print("Balance:", self.__balance)
        else:
            print("Incorrect PIN!")


a = ATM(12345, 5000, "1234")

a.check_balance()
a.withdraw()
a.deposit()
a.change_pin()
a.check_balance()
a.display_details()
````
