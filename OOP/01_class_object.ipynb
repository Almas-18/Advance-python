# OOP in Python

## Class
A class is a blueprint for creating objects.

## Object
An object is an instance of a class.

## Constructor
A constructor is automatically called when an object is created.




# 1. Class
A class is a blueprint or template used to create objects.
It defines properties and methods.

# 2. Object
An object is an instance of a class.
It can access the properties and methods defined in that class.

## Example:

```python
class Student:
    name = 'abcd'

s1 = Student()
print(s1.name)

s2 = Student()
print(s2.name)
```

**Output:**
```text
abcd
abcd
```

**Explanation:**
- `Student` → Class
- `name` → Class variable
- `s1` and `s2` → Objects
- `print(s1.name)` → Displays the name of the first object.
- `print(s2.name)` → Displays the name of the second object.


# 4. Constructor in Python

## Definition
A constructor is a special method that is automatically called when an object is created.

## Types of Constructors
1. Default Constructor – Takes no extra parameters.
2. Parameterized Constructor – Takes parameters to initialize object values.

## Constructor in Python
In Python, we use `__init__()` as a constructor.

## Rules of a Constructor
1. Constructor name: Written as `__init__()`. It does not need to match the class name.
2. No return type: It has no declared return type. `__init__()` must return `None`.
3. Automatic calling: It is called automatically when an object is created.
4. Special method: It is written with double underscores before and after `init`.

````python
class Student:
    # Data members
    RollNo=1
    name='riya'
# OBJECT
s1=Student()
print('Roll no is::',s1.RollNo)
print('Name is::',s1.name)
````

````python
class Student:
    # MEMBER FUNCTION:A function defined inside a class.
    def show(self): #self: Refers to the current object.
        print("This is a student class")
    def display(self):
        print('I am in display class')
# object
s1=Student()
s1.show()
s1.display()
s2=Student()
s2.display()
````

````python
class Employee:
    id=1
    name='ram'
    designation='devloper'
    dept='cs'
e=Employee()
print('Emp id is::',e.id)
print("emp  name is::",e.name)
print("emp designation is::",e.designation)
print("Emp department is::",e.dept)

e1=Employee()
print('Emp id is::',e1.id)
print("emp  name is::",e1.name)
print("emp designation is::",e1.designation)
print("Emp department is::",e1.dept)
````

````python
class Employee:
    def accept(self,name,salary):
        self.name=name
        self.salary=salary
    def display(self):
        print('Name is::',self.name)
        print("salary is::",self.salary)
e1=Employee()
e2=Employee()

e1.accept('Ram',3000)
e2.accept('raj',4000)
e1.display()
e2.display()
````

````python
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def display(self):
        print('Name is:',self.name)
        print('salary is::',self.salary)
e1=Employee('ram',3005)
e2=Employee('raj',5000)
e1.display()
e2.display()
````

````python
class Bank:
    def __init__(self,balance):
        self.balance=balance
    def display_bal(self):
        print('current balance is::',self.balance)
b=Bank(500)
b.display_bal()
b1=Bank(5000)
b1.display_bal()
````

````python
class EvenOdd:
    def check(self,num):
        if num%2==0:
            print('Even')
        else:
            print('odd')
n=EvenOdd()
n.check(8)
n1=EvenOdd()
n1.check(9)
````

````python
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
p1=product('Laptop',5000)
print('Name is::',p1.name,'\t','price is::',p1.price)

p2=product('Headphone',500)
print(p2.name,p2.price)
        
````

````python
# create class nME COURSES
class courses:
    def __init__(self,cid,cname,duration):
        self.cid=cid
        self.cname=cname
        self.duration=duration
c1=courses('1','python',60)
print('Course id:',c1.cid,'\t','course name:',c1.cname,'\t',c1.duration)
c2=courses('2','java','6 month')
print('Course id:',c2.cid,'\t','course name:',c2.cname,'\t',c2.duration) 
c3=courses('3','sql','3 month')
print('Course id:',c3.cid,'\t','course name:',c3.cname,'\t',c3.duration) 
````
