# Abstraction in Python

## 1. What is Abstraction?

**Definition:** Abstraction means hiding internal details and showing only essential features to the user.

**Example:**
- We use a mobile phone to make calls without knowing its internal circuits.
- Similarly, we use a function without knowing how it works internally.

## 2. Why do we use Abstraction?

1. **Reduces complexity** – Makes things easier to understand.
2. **Improves readability** – Makes code easier to read.
3. **Hides internal details** – Hides unnecessary implementation details.
4. **Promotes reusability** – Allows us to reuse code.

## 3. Simple Example

```python
class Car:
    def start(self):
        print("Car started")

car = Car()
car.start()
```

**Output:**
```text
Car started
```

**Explanation:**
- `start()` is used to start the car.
- We call `car.start()` without knowing all the internal steps.
- We only use the feature we need.

## 4. Abstraction Using ABC

Python provides the `abc` module to implement abstraction.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```

**Output:**
```text
Dog barks
```

**Explanation:**
- `ABC` – Used to create an abstract class.
- `@abstractmethod` – Defines a method that the child class must implement.
- `Dog` – Implements the `sound()` method.

## 5. Important Points

- Abstraction hides implementation details.
- It shows only essential features.
- It can be implemented using abstract classes and methods.
- An abstract class cannot be instantiated while it has unimplemented abstract methods.

**Interview Answer:** Abstraction in Python is an OOP concept that hides implementation details and shows only essential features to the user.

## 6. Rules of Abstract Class

1. An abstract class is created using `ABC`.
2. An abstract method is created using `@abstractmethod`.
3. An abstract class cannot be used to create an object if it has unimplemented abstract methods.
4. The child class must implement all abstract methods before its object can be created.

**Easy example:**

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Bark")

d = Dog()
d.sound()
```

**Output:**
```text
Bark
```

**Remember:** The parent class defines *what to do*, and the child class defines *how to do it*.
