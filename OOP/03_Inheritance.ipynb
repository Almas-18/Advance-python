# Inheritance in Python

## Definition
Inheritance is the process of acquiring properties and methods from a parent class to a child class.

- **Parent Class:** Also called the Base class.
- **Child Class:** Also called the Derived class.

## Types of Inheritance

### 1. Single Inheritance
One parent class and one child class.

### 2. Multiple Inheritance
One child class inherits from multiple parent classes.

### 3. Multilevel Inheritance
A class inherits from a child class, forming a chain of inheritance.

### 4. Hierarchical Inheritance
Multiple child classes inherit from one parent class.

### 5. Hybrid Inheritance
A combination of two or more types of inheritance.

````python
import math

class Shape:
    def sh(self):
        self.sides=int(input('Enter number of sides:'))
class Rectangle(Shape):
    def rect(self):
        self.length=float(input('Enter length:'))
        self.breadth=float(input('Enter breadth:'))
        area=self.length*self.breadth
        print('Rectangle area is:',area)
class triangle(Shape):
    def tri(self):
       self.base=float(input('Enter base:'))
       self.height=float(input('Enter height'))
       area=0.5*self.base*self.height
       print('Triangle area is:',area)
class circle(Shape):
    def circle(self):
        self.radius=float(input("Enter radius:"))
        area=math.pi*self.radius*self.radius        
        
        print('Circle area is:',area)
class pentagon(Shape):
    def pent(self):
        self.side=float(input('Enter side len:'))
        area=(5*self.side*self.side)/(4*math.tan(math.pi/5))
        print('pentagon area is:',area)
r=Rectangle()
r.sh()
r.rect()

t=triangle()
t.sh()
t.tri()

c=circle()
c.sh()
c.circle()

p=pentagon()
p.sh()
p.pent()
````

# Python `super()` Function

## 1. Definition

The `super()` function in Python is used to call methods or access functionality from the parent class inside a child class.

It is commonly used in inheritance to call the parent class constructor (`__init__()`).

## 2. Syntax

```python
super().method_name()
```

To call the parent class constructor:

```python
super().__init__(arguments)
```

## 3. Example

```python
class Parent:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name:", self.name)


class Child(Parent):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

    def show(self):
        super().display()
        print("Marks:", self.marks)


obj = Child("Almas", 90)
obj.show()
```

**Output:**

```text
Name: Almas
Marks: 90
```

## 4. Advantages of `super()`

1. **Code reusability:** Reuses the parent class code.
2. **Avoids duplication:** No need to rewrite the parent constructor logic.
3. **Easy maintenance:** Makes code easier to maintain.
4. **Supports multiple inheritance:** Helps call methods according to Python's Method Resolution Order (MRO).

## 5. How Does `super()` Work?

- `Parent` is the parent class.
- `Child` inherits from `Parent`.
- `super().__init__(name)` calls the parent constructor and initializes the name.
- `super().display()` calls the parent class's `display()` method.
- The child class also displays the student's marks.

## 6. Interview Question: Why Do We Use `super()` Instead of Directly Calling the Parent Class?

**Answer:**

1. **Easy maintenance:** `super()` follows Python's inheritance rules and makes code easier to maintain.
2. **Less code changes:** If the inheritance structure changes, `super()` can continue to work without hardcoding a specific parent class name, provided the Method Resolution Order (MRO) remains compatible.
3. **Multiple inheritance:** `super()` follows the MRO, allowing parent methods to be called in the correct order.

## 7. Interview Answer: What Is `super()`?

The `super()` function in Python is used to call methods of the parent class from a child class. It improves code reusability, avoids code duplication, and is commonly used to call the parent constructor in inheritance.

````python
#EXAM THEORY -->WHAT IS SUPER()IN PYTHON
class A:
    def display(self):
        print("Class A")

class B(A):
    def show(self):
        super().display()
        print("Class B")

obj = B()
obj.show()
````
