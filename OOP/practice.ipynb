```python
# PROJECT 1: EMPLOYEE PAYROLL MANAGEMENT SYSTEM
class Employee:
    def __init__(self,empid,empname,dept,basicsal):
        self.empid=empid
        self.empname=empname
        self.dept=dept
        self.basicsal=basicsal
    def display_detail(self):
        print('Employee ID:',self.empid)
        print("Employee Name:",self.empname)
        print("Employee department:",self.dept)
        print("Employee basic salary:",self.basicsal)
    
    def calculate_hra(self):
        return self.basicsal*0.20
    def calculate_DA(self):
        return self.basicsal*0.10
    def calculate_pf(self):
        return self.basicsal*0.12
    def calculate_gross_salary(self):
        return self.basicsal+self.calculate_hra()+self.calculate_DA()
    def calculate_net_salary(self):
        return self.calculate_gross_salary() - self.calculate_pf()
         
    def display_salary_details(self):
        print("HRA:",self.calculate_hra())
        print("DA:",self.calculate_DA())
        print("pf:",self.calculate_pf())
        print("gross salary:",self.calculate_gross_salary())
        print("net salary:",self.calculate_net_salary())
e1=Employee(101,'almas','entc',30000)
e1.display_detail()
print('\nsalary details:')
e1.display_salary_details()
        
```

```python
'''
Q5
Car Rental System
Create a Car class with a constructor to store the car name, 
rent per day, and number of days. Take input from the user and 
create an object to calculate and display the total rent.
'''
class car:
    def __init__(self,name,rent,day):
        self.name=name
        self.rent=rent
        self.day=day
    def total(self):
        return self.rent*self.day
name=input('Enter name of car:')
rent=float(input('Enter rent per day:'))
day=int(input('Enter day:'))
c1=car(name,rent,day)
print('=========caar details========')
print('name of car:',c1.name)
print('rent per day :',c1.rent)
print('no of days:',c1.day)
print('Total cost:',c1.total())

        
```

```python
class AtmAcoount:
    def __init__(self,number,name,balance):
        self.number=number
        self.name=name
        self.balance=balance
    def displaydetails(self):
        print('Account number:',self.number)
        print('Customer Name:',self.name)
        print('Balance :',self.balance)
    def deposit(self,damount):
        self.balance+=damount
        print('Amount Deposited:',damount)
    def withdraw(self,wamount):
        if wamount<=self.balance:
            self.balance-=wamount
            print('Amount Withdraw:',wamount)
        else:
            print('Insufficient Balance')
    def updated_balance(self):
        print('updated balance:',self.balance)
    
a=AtmAcoount(1001,'sanam',50000)
a.displaydetails()
a.deposit(2000)
a.withdraw(200)
a.updated_balance()

        
```

```python
class RestaurantOrder:
    def __init__(self,id,name,food,price,quantity):
        self.id=id
        self.name=name
        self.food=food
        self.price=price
        self.quantity=quantity
    def display(self):
        total=self.price*self.quantity
        if total>2000:
            discount=total*10/100
        else:
            discount=0
        finalamount=total-discount
        print("\nOrder ID:", self.id)
        print("Customer:", self.name)
        print("Food:", self.food)
        print("Total:", total)
        print("Discount:", discount)
        print("Final Amount:", finalamount)
r1=RestaurantOrder(101,'saniya','pizza',2000,5)
r1.display()
        
```

```python
class Course:
    def __init__(self,cid,cname,instructor,duration,fees):
        self.cid=cid
        self.cname=cname
        self.instructor=instructor
        self.duration=duration
        self.fees=fees
    def display(self):
        print(self.cid,self.cname,self.instructor,self.duration,self.fees)
c1 = Course(1, "Python", "Amit", "3 months", 10000)
c2 = Course(2, "Java", "Neha", "4 months", 15000)
c3 = Course(3, "SQL", "Ravi", "2 months", 8000)       
courses=[c1,c2,c3]
for c in courses:
    c.display()
highest=max(courses,key=lambda c:c.fees)
print('Highest fees course:',highest.cname)
print('Fees',highest.fees)
```

```python

class BankAccount:
    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def display(self):
        print("\nAccount Number:", self.account_number)
        print("Customer Name:", self.customer_name)
        print("Balance:", self.balance)


a1 = BankAccount(1001, "Amit", 5000)
a2 = BankAccount(1002, "Neha", 8000)

a1.deposit(2000)
a2.deposit(1500)

a1.display()
a2.display()
```
