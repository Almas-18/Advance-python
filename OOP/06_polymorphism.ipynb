# Polymorphism in Python

## 1. What is Polymorphism?

Polymorphism means **"Many Forms".**

In Python, polymorphism allows the same function or method name to behave differently depending on the object or data type.

**In simple words:** One interface – Multiple implementations.

### Example

```python
print(len("Python"))       # 6
print(len([10, 20, 30]))   # 3
print(len((1, 2, 3, 4)))   # 4
```

Here, the same `len()` function works with different data types.

---

## 2. Types of Polymorphism in Python

1. **Operator Overloading**
2. **Method Overloading**
3. **Method Overriding**
4. **Duck Typing**

---

## 2.1 Operator Overloading

**Definition:**

Operator overloading allows operators such as `+`, `-`, `*`, and `==` to perform different operations depending on the objects.

**Use:** Dunder Methods (Double Underscore Methods).

## 2.2 Method Overloading

**Definition:**

When multiple methods have the same name but different parameters, it is called Method Overloading.

**In Python, we can simulate method overloading using:**

- Default Arguments
- `*args`

## 2.3 Method Overriding

**Definition:**

When a child class provides a different implementation of a method that already exists in the parent class, it is called Method Overriding.

## 2.4 Duck Typing

**Definition:**

Python focuses on behaviour, not object type.

**In simple words:** If an object has the required method, Python allows us to use it, regardless of its class.

### Example

```python
class Bird:
    def fly(self):
        print("Bird can fly")


class Airplane:
    def fly(self):
        print("Airplane flies")


def make_fly(obj):
    obj.fly()


b = Bird()
a = Airplane()

make_fly(b)
make_fly(a)

No inheritance required ==>just same methond name
## Explanation
- Bird and Airplane are two different classes.
- Both classes have the same fly() method.
- make_fly(obj) accepts both objects.
- Python focuses on the behaviour (fly method), not the object type.
Conclusion: Duck Typing means Python focuses on what an object can do rather than what type of object it is.


## 3. Quick Revision

| Sr. No. | Type of Polymorphism | Main Concept | 
|:---:|---|---|
| 1 | Operator Overloading | Dunder Methods (`__add__`, `__sub__`) |
| 2 | Method Overloading | Default Arguments and `*args` |
| 3 | Method Overriding | Child class redefines the parent class method |
| 4 | Duck Typing | Focuses on behaviour, not object type |

1. **Compile-Time Polymorphism**
2. **Run-Time Polymorphism (Dynamic Polymorphism)**

- It is also called static polymorphism.
- The method to be executed is determined before program execution.
- Method overloading is a common example in languages like Java and C++.
- Python does not directly support traditional compile-time method overloading.

### 2. Run-Time Polymorphism

- It is also called dynamic polymorphism.
- The method to be executed is determined during program execution.
- Python achieves runtime polymorphism through method overriding and duck typing.

- A child class provides its own implementation of a parent class method.

### 4. Duck Typing
- Achieved by focusing on **behaviour rather than object type**.

# Polymorphism in Python

Polymorphism means **one name, many forms**.

It means the same method can perform different tasks in different situations.

## 1. Compile-time Polymorphism (Static Polymorphism)

### Definition

Compile-time polymorphism means deciding which method to use before the program runs.

Python does not directly support compile-time polymorphism like Java and C++. But we can achieve similar behaviour using default arguments and `*args`.

### Example: Using Default Arguments

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

obj = Calculator()

print(obj.add(10, 20))
print(obj.add(10, 20, 30))
```

### Output

```text
30
60
```

### Explanation

- First, we pass 2 numbers, so it adds 10 + 20.
- Next, we pass 3 numbers, so it adds 10 + 20 + 30.
- The same method `add()` works with different numbers of arguments.

### Advantages

1. We can use the same method for different numbers of arguments.
2. Makes code reusable.
3. Reduces the need to write separate methods.

---

## 2. Run-time Polymorphism (Dynamic Polymorphism)

### Definition

Run-time polymorphism means the method is selected when the program is running, depending on which object is calling it.

In Python, it is achieved using **Method Overriding**.

### Example: Method Overriding

```python
class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

# Creating objects
d = Dog()
c = Cat()

d.sound()
c.sound()
```

### Output

```text
Dog barks
Cat meows
```

### Explanation

- `Animal` is the parent class.
- `Dog` and `Cat` are child classes.
- Both child classes have the same method `sound()`.
- But each class gives a different output.
- When we call `sound()`, Python checks which object's method should run.

### Advantages

1. The same method can perform different tasks.
2. Makes code flexible.
3. Helps in code reusability.
4. Makes code easy to manage.

---
## Difference Between Compile-time Polymorphism and Run-time Polymorphism

### 1. Decision Time

- **Compile-time Polymorphism:** The method is decided before the program runs.
- **Run-time Polymorphism:** The method is decided while the program is running.

### 2. Technique Used

- **Compile-time Polymorphism:** Uses Method Overloading (simulated using default arguments or `*args` in Python).
- **Run-time Polymorphism:** Uses Method Overriding.

### 3. Speed

- **Compile-time Polymorphism:** Generally faster because the method is already decided before execution.
- **Run-time Polymorphism:** May be slightly slower because the method is selected during program execution.

### 4. Flexibility

- **Compile-time Polymorphism:** Less flexible because the method is decided earlier.
- **Run-time Polymorphism:** More flexible because different objects can perform different tasks using the same method.
### 5. Binding

- **Compile-time Polymorphism:** Uses Early Binding (Static Binding). The method is connected before the program runs.
- **Run-time Polymorphism:** Uses Late Binding (Dynamic Binding). The method is selected while the program is running.

### 6. Python Support

- **Compile-time Polymorphism:** Python does not support true compile-time polymorphism. However, we can simulate method overloading using default arguments and `*args`.
- **Run-time Polymorphism:** Python supports runtime polymorphism using Method Overriding.

### Remember

- **Early Binding** = Decided before execution.
- **Late Binding** = Decided during execution.

### 7. Example

- **Compile-time Polymorphism:** Adding two or three numbers using the same method.
- **Run-time Polymorphism:** Dog barks and Cat meows using the same `sound()` method.

### 6. Python Support

- **Compile-time Polymorphism:** Python does not directly support traditional compile-time polymorphism.
- **Run-time Polymorphism:** Python supports runtime polymorphism using method overriding.

Duck Typing in Python (In Detail)

1. Definition

Duck typing is a concept in Python where the type or class of an object is less important than what the object can do.

In simple words, Python does not check which class an object belongs to. It checks whether the object has the required methods or behaviour.

2. Famous Phrase

"If it walks like a duck and quacks like a duck, then it must be a duck.

3. Real-life Example 🦆
Think about a duck:
- A duck can swim.
- A fish can also swim.
- Python doesn't care whether they are a duck or fish. If both can perform the swim() action, we can call that method.

3. Python Example
class Duck:
    def sound(self):
        print("Quack Quack")

class Dog:
    def sound(self):
        print("Bhow Bhow")

def make_sound(animal):
    animal.sound()

# Creating objects
d = Duck()
dog = Dog()

make_sound(d)
make_sound(dog)

Explanation

We created two classes: Duck and Dog.

Both classes have the same method sound().

We created a function make_sound().

This function accepts any object and calls its sound() method.

Python does not check whether the object is a Duck or Dog.

It only checks whether the object has the sound() method.

This is called Duck Typing.
5. What Happens If the Method Is Missing?

class Cat:
    def walk(self):
        print("Cat is walking")

def make_sound(animal):
    animal.sound()

c = Cat()
make_sound(c)

Output

AttributeError: 'Cat' object has no attribute 'sound'

Explanation

The Cat class does not have a sound() method. So Python gives an AttributeError when we try to call it.

6. Advantages of Duck Typing

Flexibility: We can use different objects with the same function.

Code Reusability: We can use one function for different classes.

Less Code: No need to check the object's class.

Easy to Use: We only need the required method or behaviour.

7. Interview Question

Q. What is Duck Typing in Python?

Duck typing is a concept in Python where we focus on what an object can do instead of checking its class or type. If an object has the required method, we can use it.

Remember:

Duck Typing = Don't check the object's type; check what the object can do.
